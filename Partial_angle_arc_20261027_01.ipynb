```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create figure with a transparent background
fig, ax = plt.subplots(subplot_kw={'projection': 'polar'}, figsize=(6, 6))
fig.patch.set_alpha(0.0)
ax.patch.set_alpha(0.0)

# Set 0° at the top ('N') 
# Counter-clockwise direction: negative numbers to the right, positive to the left
ax.set_theta_zero_location('S')
ax.set_theta_direction(1) ## takes values of: 1 or -1

# Define the arc range from -90° (right) to +20° (left)
theta_min = np.deg2rad(-90)
theta_max = np.deg2rad(20)
ax.set_thetamin(np.rad2deg(theta_min))
ax.set_thetamax(np.rad2deg(theta_max))

# Set radial limits
ax.set_rmin(0.0)
ax.set_rmax(1.0)

# Configure ticks every 5 degrees
tick_angles = np.arange(-90, 21, 10)
ax.set_xticks(np.deg2rad(tick_angles))

# Format labels with larger, bold font
tick_labels = [f"{int(deg)}°" for deg in tick_angles]
ax.set_xticklabels(tick_labels, fontsize=14, fontweight='bold')

# Make ticks visible and add padding to prevent text overlap
ax.tick_params(
    axis='x', 
    pad=12,         # Pushes the text labels further away from the arc line
    length=6,       # Makes the tick marks visible (length in points)
    width=1.5,      # Tick mark thickness
    direction='out' # Points ticks outward from the arc
)

# Clean up styling: remove radial gridlines and inner labels, keep outer arc spine
ax.set_yticklabels([])
ax.grid(False)
ax.spines['polar'].set_visible(True)
ax.spines['polar'].set_color('black')
ax.spines['polar'].set_linewidth(2)

# Save with a transparent background
plt.savefig('protractor_arc.svg', bbox_inches='tight', transparent=True)
plt.show()
```
