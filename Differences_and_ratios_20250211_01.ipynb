## The change in a ratio depends on the starting ratio and change of each component of the ratio

starting from a ratio that is different from 1:1 indroduces unintuitive effects when each of the operands of the ratio are variable

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
changes_arr_0 = np.linspace(-0.99,5,50)
out_arr = np.zeros((50,50))
```

```python
#out_2 = np.meshgrid([changes_arr_0, changes_arr_0])
#out_2
```

```python

glu_start = 5
gaba_start = 1

for idx_row, gaba_row_i in enumerate(changes_arr_0):
    for idx_col, glu_col_j in enumerate(changes_arr_0):
        out_arr[idx_row, idx_col] = ((glu_start+glu_col_j) / (gaba_start+gaba_row_i)) - (glu_start / gaba_start)
        
```

```python
fig, ax = plt.subplots(1,1)
#ax.imshow(out_arr)
ax_obj = ax.imshow(out_arr, cmap='seismic', vmin=-15, vmax=15, origin='lower', extent=[-0.99,5,-0.99,5])
#ax_obj = ax.matshow(out_arr, cmap='seismic', vmin=-10, vmax=10, origin='lower')

ax.plot(changes_arr_0, changes_arr_0, color='k', ls='--', lw=0.8)
ax.hlines(0, -0.99, 5, color='green', ls='--', lw=0.6)
ax.vlines(0, -0.99, 5, color='green', ls='--', lw=0.6)


fig.colorbar(ax_obj, ax=ax, label='difference of ratios \n (Glu/GABA Follow - Glu/GABA BL)')
ax.set_title('Starting ratio Glu/GABA: Glu=5.0, GABA=1.0')
ax.set_xlabel('Change in Glu')
ax.set_ylabel('Change in GABA')
#ax.set_xticks(changes_arr_0)
#ax.set_yticks(changes_arr_0)
plt.show()
```
